# 📸 ParkKaro - YOLOv8n Parking Sign Scanner Fine-Tuning

Fine-tuning YOLOv8n (Nano) on Indian parking & traffic signs.
Target output: `backend/models/sign_detector.pt`

### Supported Classes:
- 0: `No Parking`
- 1: `Tow Away Zone`
- 2: `Paid Parking`
- 3: `Free Parking`
- 4: `No Stopping`

```python
# Step 3: Train YOLOv8n (Lightweight for fast CPU/edge inference)
from ultralytics import YOLO

# Load pre-trained nano model
model = YOLO('yolov8n.pt')

# Train for 25 epochs
results = model.train(
    data='data.yaml',
    epochs=25,
    imgsz=640,
    batch=16,
    name='parkkaro_sign_detector'
)
```
