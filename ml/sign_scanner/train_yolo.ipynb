# 📸 ParkKaro - YOLOv8n Parking & Traffic Sign Scanner

Fine-tuning YOLOv8n (Nano) on the Indian Parking & Traffic Signs dataset (`ParkKaro.v1i.yolov8`).
Target output: `backend/models/sign_detector.pt`

### Dataset:
- Classes: 85 Indian Traffic & Parking Sign categories (including NO_PARKING, NO_STOPPING, etc.)
- Config: `data.yaml`

```python
# Step 1: Verify Ultralytics & PyTorch CUDA
import torch
print(f"PyTorch: {torch.__version__} | CUDA Available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"Device: {torch.cuda.get_device_name(0)}")
```

```python
# Step 2: Train YOLOv8n
from ultralytics import YOLO
import shutil
from pathlib import Path

# Load pre-trained nano model
model = YOLO('yolov8n.pt')

# Train for 25 epochs
results = model.train(
    data='data.yaml',
    epochs=25,
    imgsz=640,
    batch=8,
    project='runs',
    name='parkkaro_sign_detector',
    exist_ok=True
)
```

```python
# Step 3: Export best weights to backend/models/sign_detector.pt
best_pt = Path('runs/parkkaro_sign_detector/weights/best.pt')
dest_pt = Path('../../backend/models/sign_detector.pt')
if best_pt.exists():
    dest_pt.parent.mkdir(parents=True, exist_ok=True)
    shutil.copy2(best_pt, dest_pt)
    print(f"✅ Exported {best_pt} to {dest_pt} ({dest_pt.stat().st_size / (1024*1024):.2f} MB)")
else:
    print(f"⚠️ Weights not found at {best_pt}")
```
